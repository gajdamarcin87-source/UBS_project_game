```python
import pandas as pd
from pathlib import Path

csv_file_easy = Path("quiz_easy_100.csv")
csv_file_medium = Path("quiz_medium_100.csv")
csv_file_hard = Path("quiz_hard_100.csv")

easy_df = pd.read_csv(csv_file_easy)
medium_df = pd.read_csv(csv_file_medium)
hard_df = pd.read_csv(csv_file_hard)
```

```python
while True:
    try:
        player_count = int(input("How many players? (1–4): "))
        if 1 <= player_count <= 4:
            break
        print("Choose a number from 1 to 4.")
    except ValueError:
        print("Enter a number.")

players = [
    {"name": input(f"Name for player {i + 1}: ").strip() or f"Player {i + 1}",
     "score": 0}
    for i in range(player_count)
]
```

```python
import time

TIME_LIMIT = 30
max_questions = 10

def normalize_answer(value):
    return str(value).strip().upper()

difficulty_map = {
    "EASY": easy_df,
    "MEDIUM": medium_df,
    "HARD": hard_df,
}

for player in players:
    print(f"\n--- {player['name']}'s turn ---")

    quiz_difficulty = normalize_answer(
        input("Select difficulty (EASY, MEDIUM, HARD): ")
    )

    if quiz_difficulty not in difficulty_map:
        print("Invalid difficulty. Skipping this player's turn.")
        player["score"] = 0
        continue

    selected_df = difficulty_map[quiz_difficulty]
    questions = selected_df.sample(n=min(max_questions, len(selected_df)))

    correct_score = 0
    incorrect_score = 0

    for question_number, (_, question) in enumerate(
        questions.iterrows(), start=1
    ):
        print(f"\nQuestion {question_number}: {question['Question']}")

        for option in ["A", "B", "C", "D"]:
            print(f"{option}: {question[option]}")

        start_time = time.monotonic()
        answer = normalize_answer(input("Your answer (A/B/C/D): "))
        elapsed = time.monotonic() - start_time

        if elapsed > TIME_LIMIT:
            print(f"Time's up! You took {elapsed:.0f}s.")
            break

        while answer not in {"A", "B", "C", "D"}:
            answer = normalize_answer(
                input("Invalid input. Choose A, B, C, or D: ")
            )

        correct_answer = normalize_answer(question["Correct_Answer"])

        if answer == correct_answer:
            print("Correct!")
            correct_score += 1
        else:
            print(f"Incorrect. The correct answer is {correct_answer}.")
            incorrect_score += 1

    player["score"] = correct_score
    print(f"Score: {correct_score}; incorrect answers: {incorrect_score}")

print("\n--- Final scores ---")
for player in sorted(players, key=lambda p: p["score"], reverse=True):
    print(f"{player['name']}: {player['score']} points")
```
