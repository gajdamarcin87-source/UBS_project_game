```python
import pandas as pd
from pathlib import Path

csv_file_easy = Path("quiz_easy_100.csv")
csv_file_medium = Path("quiz_medium_100.csv")
csv_file_hard = Path("quiz_hard_100.csv")

easy_df = pd.read_csv(csv_file_easy)
medium_df = pd.read_csv(csv_file_medium)
hard_df = pd.read_csv(csv_file_hard)
```

```python
def normalize_answer(value):
    return str(value).strip().upper()


difficulty_map = {
    "EASY": easy_df,
    "MEDIUM": medium_df,
    "HARD": hard_df,
}

max_questions = 10

quiz_difficulty = normalize_answer(
    input("Select difficulty (EASY, MEDIUM, HARD): ")
)

if quiz_difficulty not in difficulty_map:
    print("Invalid difficulty. Please choose EASY, MEDIUM, or HARD.")
else:
    selected_df = difficulty_map[quiz_difficulty]
    questions = selected_df.sample(n=min(max_questions, len(selected_df)))
    correct_score = 0
    incorrect_score = 0

    for question_number, (_, question) in enumerate(questions.iterrows(), start=1):
        print(f"\nQuestion {question_number}: {question['Question']}")

        for option in ["A", "B", "C", "D"]:
            print(f"{option}: {question[option]}")

        answer = normalize_answer(input("\nYour answer (A/B/C/D): "))

        while answer not in {"A", "B", "C", "D"}:
            print("Incorrect answer. Choose from A, B, C or D.")
            answer = normalize_answer(input("\nYour answer (A/B/C/D): "))

        correct_answer = normalize_answer(question["Correct_Answer"])

        if answer == correct_answer:
            print("Correct!")
            correct_score += 1
        else:
            print(f"Incorrect. The correct answer is {correct_answer}.")
            incorrect_score += 1
    print(f"\nQuiz complete! Your score: {correct_score}/{len(questions)}")
    print(f"Incorrect answers: {incorrect_score}/{len(questions)}")
```
