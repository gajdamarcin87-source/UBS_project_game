```python
import pandas as pd
from pathlib import Path

csv_file_easy = Path("quiz_easy_100.csv")
csv_file_medium = Path("quiz_medium_100.csv")
csv_file_hard = Path("quiz_hard_100.csv")

easy_df = pd.read_csv(csv_file_easy)
medium_df = pd.read_csv(csv_file_medium)
hard_df = pd.read_csv(csv_file_hard)
```

```python
quiz_dificulty = int(input("Select quiz difficulty (1: Easy, 2: Medium, 3: Hard): "))
```

```python
difficulty_map = {
    1: easy_df,
    2: medium_df,
    3: hard_df,
}

if quiz_dificulty in difficulty_map:
    selected_df = difficulty_map[quiz_dificulty]
    question = selected_df.sample(n=1).iloc[0]

    print(f"\n{question['Question']}")
    for option in ["A", "B", "C", "D"]:
        print(f"{option}: {question[option]}")

    answer = input("\nYour answer (A/B/C/D): ").strip().upper()
    correct_answer = str(question["Correct_Answer"]).strip().upper()

    if answer == correct_answer:
        print("Correct!")
    else:
        print(f"Incorrect. The correct answer is {correct_answer}.")
else:
    print("Invalid difficulty. Please choose 1, 2, or 3.")
```
